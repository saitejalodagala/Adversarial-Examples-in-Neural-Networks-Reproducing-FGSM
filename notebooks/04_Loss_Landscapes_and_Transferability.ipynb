## 1. 1D Loss Landscape Trajectory
By computing the loss along the line $L(oldsymbol{	heta}, \mathbf{x} + \gamma \cdot oldsymbol{\delta}, y)$, we can directly visualize the smoothness and linearity of the decision space.


```python
import os
import sys
import torch
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath(".."))

from adv_studio.models import get_model
from adv_studio.data import get_sample_digits
from adv_studio.attacks import FGSMAttack
from adv_studio.evaluation import compute_loss_landscape_1d, compute_loss_landscape_2d
from adv_studio.visualization import plot_loss_landscape_1d, plot_loss_landscape_2d

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = get_model("simple_cnn", pretrained_path="../checkpoints/mnist_cnn.pth", device=device)

sample_imgs, sample_lbls = get_sample_digits(data_dir="../DATA", count=1)
sample_imgs, sample_lbls = sample_imgs.to(device), sample_lbls.to(device)

attack = FGSMAttack(model, epsilon=0.25, device=device)
adv_img = attack.generate(sample_imgs, sample_lbls)

landscape_data = compute_loss_landscape_1d(model, sample_imgs[0:1], sample_lbls[0:1], adv_img[0:1])
fig = plot_loss_landscape_1d(landscape_data)
plt.show()
```

## 2. 2D Loss Surface Contour Map


```python
landscape_2d = compute_loss_landscape_2d(model, sample_imgs[0:1], sample_lbls[0:1], adv_img[0:1])
fig = plot_loss_landscape_2d(landscape_2d)
plt.show()
```
